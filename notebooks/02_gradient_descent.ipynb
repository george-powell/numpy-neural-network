This notebook builds on the previous, instead assuming the true value of m and c are unknown and not algebraically solvable. Hence, we are left instead to rely on stochastic gradient descent to approximate their true values. 

Stochastic gradient descent is used for updating weights and biases in a neural network model during training.

```python
# importing dependencies
import numpy as np
import matplotlib.pyplot as plt
from numpy import random
```

```python
# initial conditions
true_m = 0.5
true_c = 1

n = 50
x_range = 10
sim = 1

x = np.linspace(0, x_range, n)
y = true_m * x + true_c
```

```python
# initial conditions
m_est = 0
c_est = 0

loss_tracker = []

lr = 0.01

for epoch in range(500):
    y_est = m_est * x + c_est
    loss = np.mean((y - y_est)**2)

    loss_tracker.append(loss)

    dm = np.mean(-2 * x *(y - y_est))
    dc = np.mean(-2 * (y - y_est))

    
    m_est -= lr * dm
    c_est -= lr * dc

```

### Visualisation

```python
# plots loss on a logarithmic y-axis.
plt.figure()
plt.semilogy(loss_tracker)
plt.ylabel("logged loss of regression line estimate")
plt.xlabel("epoch")
plt.show()
```

```python
# Compare true and estimated parameters. 
plt.figure(figsize=(10,6))

plt.scatter(x, y, label=f"exact: m={true_m}, c={true_c}")

bfl_x = np.linspace(0,10,100)
bfl_y = m_est * bfl_x + c_est
plt.plot(bfl_x, bfl_y, label=f"estimated: m={m_est:.2f}, c={c_est:.2f}")
plt.title("Linear Regression by gradient descent")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
